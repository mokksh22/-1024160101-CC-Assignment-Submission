Q.1 Write a program to create a NumPy 1D-array with 5 elements and perform basic
operations like:
a) Addition of 2 in all the element  
b) Multiply 3 with all the elements
c) Divide every element by 2

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print("Original Array:", arr)

addition = arr + 2
print("After adding 2:", addition)

multiplication = arr * 3
print("After multiplying by 3:", multiplication)

division = arr / 2
print("After dividing by 2:", division)
```

Q.2 Questions on Basic NumPy Array:

a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5])

b) Find the most frequent value and their indice(s) in the following arrays:

i. x = np.array([1,2,3,4,5,1,2,1,1,1])

ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])

```python
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])

print("Original Array:", arr)

reverse = arr[::-1]
print("Reversed Array:", reverse)

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

print("Most frequent element in x:", np.bincount(x).argmax())
print("Indices of most frequent element in x:", np.where(x == np.bincount(x).argmax())[0])

print("Most frequent element in y:", np.bincount(y).argmax())
print("Indices of most frequent element in y:", np.where(y == np.bincount(y).argmax())[0])
```

Q.3 For the given 2-D array arr=np.array([10, 20, 30], [40, 50, 60], [70, 80, 90]), access

Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly
spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array,
shape, total elements, the data type of each element and total number of bytes consumed
by the array. Find the transpose of this array using reshape() attribute. Can we do the same
with T attribute?

```python
import numpy as np

Mokksh = np.linspace(10, 100, 25)

print("Array:")
print(Mokksh)

print("Dimensions:", Mokksh.ndim)
print("Total elements:", Mokksh.size)
print("Data type of each element:", Mokksh.dtype)
print("Total bytes consumed:", Mokksh.nbytes)

transpose_reshape = Mokksh.reshape(25, 1)

print("Transpose using reshape():")
print(transpose_reshape)

transpose_T = Mokksh.T

print("Using T attribute:")
print(transpose_T)

```

Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>>
with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean,
median, max, min, unique elements. Reshape the array to four rows and three columns and
name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three
columns and name it as resized_ ucs420_<your_name>>.

```python
import numpy as np

ucs420_Mokksh = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_Mokksh)

print("Mean:", np.mean(ucs420_Mokksh))
print("Median:", np.median(ucs420_Mokksh))
print("Maximum:", np.max(ucs420_Mokksh))
print("Minimum:", np.min(ucs420_Mokksh))
print("Unique Elements:", np.unique(ucs420_Mokksh))

reshaped_ucs420_Mokksh = ucs420_Mokksh.reshape(4, 3)
print("Reshaped Array:")
print(reshaped_ucs420_Mokksh)

resized_ucs420_Mokksh = np.resize(ucs420_Mokksh, (2, 3))
print("Resized Array:")
print(resized_ucs420_Mokksh)
```
